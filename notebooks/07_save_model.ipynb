```python
# ---- Save our trained model to a file so the app can load it instantly ----
import pandas as pd                                      # data handling
import sqlite3                                            # database access
import joblib                                             # joblib: saves/loads models to disk
from sklearn.linear_model import LogisticRegression      # our model

# Load the features from the database
conn = sqlite3.connect(r'C:\Users\User\Desktop\football-xg-model\data\football_xg.db')
df = pd.read_sql("SELECT * FROM shot_features", conn)

# Set up features and target
X = df[['distance', 'angle', 'is_header', 'is_penalty', 'is_open_play']]
y = df['goal']

# Train the model on all the data
model = LogisticRegression(max_iter=1000)   # create the model
model.fit(X, y)                             # train it

# Save the trained model to a file called 'xg_model.pkl' in the repo folder
# joblib.dump writes the whole trained model to disk so we can reload it later
joblib.dump(model, r'C:\Users\User\Desktop\football-xg-model\xg_model.pkl')

print("Model saved! The app can now load it instantly.")
```
